# Create Audio Augmentation

This notebook creates augmented WAV files for selected training experiments.

Audio augmentation is used to increase the variability of the training data. Validation and test data are kept unchanged in the later feature extraction steps.

```python
import sys
from pathlib import Path

current_path = Path.cwd()

for parent in [current_path] + list(current_path.parents):
    if (parent / "src").exists():
        PROJECT_ROOT = parent
        break
else:
    raise FileNotFoundError("Could not find project root containing 'src' folder.")

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
```

```python
import pandas as pd

from src.config import (
    PROCESSED_TRACKS_CSV,
    WAV_AUDIO_DIR,
    AUGMENTED_WAV_AUDIO_DIR,
    N_AUGMENTATIONS
)

from src.preprocessing_functions import augment_audio_files
```

## Load metadata

Only tracks with a valid track identifier and top-level genre label are used for augmentation.

```python
metadata = pd.read_csv(PROCESSED_TRACKS_CSV)

metadata = metadata[
    [
        "track_id",
        "track_genre_top"
    ]
]

metadata = metadata.dropna(subset=["track_id", "track_genre_top"]).copy()
metadata["track_id"] = metadata["track_id"].astype(str)

metadata.head()
```

## Generate augmented audio files

For each valid WAV file, several augmented versions are created.  
The generated files are stored in a separate output directory and can later be used for training data augmentation.
## Generate augmented audio files

The original WAV files are augmented and stored in a separate output directory.

```python
augment_audio_files(
    inputdir=str(WAV_AUDIO_DIR),
    metadata_df=metadata,
    outdir=str(AUGMENTED_WAV_AUDIO_DIR),
    n_augmentations=N_AUGMENTATIONS
)
```
