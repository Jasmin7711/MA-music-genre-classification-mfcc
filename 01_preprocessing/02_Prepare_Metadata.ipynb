# Prepare FMA Metadata

This notebook prepares the Free Music Archive metadata for the following preprocessing steps.

The relevant metadata columns are loaded and stored in a cleaned CSV file that is used by the feature extraction notebooks.

```python
import sys
from pathlib import Path

current_path = Path.cwd()

for parent in [current_path] + list(current_path.parents):
    if (parent / "src").exists():
        PROJECT_ROOT = parent
        break
else:
    raise FileNotFoundError("Could not find project root containing 'src' folder.")

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
```

```python
import pandas as pd

from src.config import (
    RAW_TRACKS_CSV,
    PROCESSED_TRACKS_CSV
)
```

## Load metadata

The metadata file is loaded from the configured dataset directory.  
Depending on the available file, the metadata may already be available as a flattened CSV file.

```python
tracks = pd.read_csv(RAW_TRACKS_CSV)

tracks.head()
```

```python
tracks["track_id"] = pd.to_numeric(
    tracks["track_id"],
    errors="coerce"
)

tracks = tracks.dropna(subset=["track_id"]).copy()
tracks["track_id"] = tracks["track_id"].astype(int)

tracks[["track_id", "track_genre_top"]].head()
```

```python
missing_genres = tracks["track_genre_top"].isna().sum()

print(f"Missing values in track_genre_top: {missing_genres}")
```

```python
tracks.to_csv(PROCESSED_TRACKS_CSV, index=False)

print(f"Prepared metadata saved to: {PROCESSED_TRACKS_CSV}")
```
